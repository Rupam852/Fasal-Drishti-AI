```python
import os
import shutil
import glob

# 1. Output Unified Folder
UNIFIED_DIR = "/kaggle/working/unified_fasal_dataset"
if os.path.exists(UNIFIED_DIR):
    shutil.rmtree(UNIFIED_DIR)
os.makedirs(UNIFIED_DIR, exist_ok=True)

base_db = "/kaggle/input/datasets/rupambairagya/fasal-drishti-ml-db"

# Helper function: fast copy/link
def add_images_to_class(src_folder, class_name):
    target_class_dir = os.path.join(UNIFIED_DIR, class_name)
    os.makedirs(target_class_dir, exist_ok=True)
    for ext in ("*.jpg", "*.JPG", "*.png", "*.PNG", "*.jpeg", "*.Jpg"):
        for img_path in glob.glob(os.path.join(src_folder, ext)):
            fname = os.path.basename(img_path)
            dst = os.path.join(target_class_dir, f"{class_name}_{fname}")
            try:
                os.symlink(img_path, dst)
            except:
                shutil.copy(img_path, dst)

print("⏳ Merging Datasets... Please wait 10-15 seconds...")

# -------------------------------------------------------------
# 1. PLANTVILLAGE (38 Standard Classes)
# -------------------------------------------------------------
pv_color = os.path.join(base_db, "plantVillage/PlantVillage-Dataset-master/raw/color")
if os.path.exists(pv_color):
    for c in os.listdir(pv_color):
        c_path = os.path.join(pv_color, c)
        if os.path.isdir(c_path):
            add_images_to_class(c_path, c)
    print("✅ 1. PlantVillage merged!")

# -------------------------------------------------------------
# 2. ICAR RICE & MAIZE (Indian Rice & Corn Diseases)
# -------------------------------------------------------------
icar_base = os.path.join(base_db, "icar_rice_Maize")
# Rice Diseases
add_images_to_class(os.path.join(icar_base, "Rice/Disease/01_Bacterial_leaf_blight"), "Rice___Bacterial_blight")
add_images_to_class(os.path.join(icar_base, "Rice/Disease/02_Brown_spot"), "Rice___Brown_spot")
add_images_to_class(os.path.join(icar_base, "Rice/Disease/04_leaf_sheath_blight"), "Rice___Sheath_blight")
add_images_to_class(os.path.join(icar_base, "Rice/Healthy"), "Rice___healthy")
# Maize Diseases
add_images_to_class(os.path.join(icar_base, "Maize/Disease/01_maydis_leaf_blight"), "Corn_(maize)___Northern_Leaf_Blight")
add_images_to_class(os.path.join(icar_base, "Maize/Disease/02_turcicum_leaf_blight"), "Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot")
add_images_to_class(os.path.join(icar_base, "Maize/Healthy"), "Corn_(maize)___healthy")
print("✅ 2. ICAR Indian Rice & Maize merged!")

# -------------------------------------------------------------
# 3. PLANTDOC (Real Indian Field Images)
# -------------------------------------------------------------
plantdoc_base = os.path.join(base_db, "plantDocdetection/PlantDoc-Object-Detection-Dataset-master")
if os.path.exists(plantdoc_base):
    for sub in ["TRAIN", "TEST"]:
        sub_dir = os.path.join(plantdoc_base, sub)
        if os.path.exists(sub_dir):
            for item in os.listdir(sub_dir):
                item_path = os.path.join(sub_dir, item)
                if os.path.isdir(item_path):
                    # Clean class name
                    clean_name = item.replace(" leaf", "").replace(" ", "_")
                    add_images_to_class(item_path, clean_name)
    print("✅ 3. PlantDoc real-field images merged!")

# -------------------------------------------------------------
# FINAL STATS
# -------------------------------------------------------------
final_classes = [d for d in os.listdir(UNIFIED_DIR) if os.path.isdir(os.path.join(UNIFIED_DIR, d))]
total_imgs = sum(len(os.listdir(os.path.join(UNIFIED_DIR, d))) for d in final_classes)

print(f"\n🎉 MEGA DATASET READY!")
print(f"📁 Path: {UNIFIED_DIR}")
print(f"🌾 Total Classes: {len(final_classes)} (Rice, Maize, Tomato, Potato, Apple etc.)")
print(f"🖼️ Total Images: {total_imgs}")
```

```python
import os
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks

# 1. 🔥 T4 GPU TENSOR CORES ON KAREIN (3x Speed Boost)
tf.keras.mixed_precision.set_global_policy('mixed_float16')

# 2. 🔥 DONO T4 GPUs KO EK SATH KAAM PAR LAGAYEIN
strategy = tf.distribute.MirroredStrategy()
print(f"Number of GPUs active: {strategy.num_replicas_in_sync}")

DATASET_DIR = "/kaggle/working/unified_fasal_dataset"
IMG_SIZE = (224, 224)
BATCH_SIZE = 128  # Bada batch size taki GPU 100% full utilize ho

# Data pipeline with parallel disk read
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical"
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical"
)

class_names = train_ds.class_names
NUM_CLASSES = len(class_names)

# In-Memory Cache + Prefetch (CPU ko free karega)
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

# 3. 🔥 MODEL WITH GPU-NATIVE AUGMENTATION
with strategy.scope():
    inputs = tf.keras.Input(shape=(224, 224, 3))
    
    # Ye Augmentation ab CPU par nahi, seedhe GPU CUDA cores par chalega!
    x = layers.RandomFlip("horizontal_and_vertical")(inputs)
    x = layers.RandomRotation(0.2)(x)
    x = layers.RandomZoom(0.15)(x)
    
    # Preprocessing (-1 to 1)
    x = layers.Rescaling(1./127.5, offset=-1)(x)
    
    base_model = tf.keras.applications.MobileNetV2(
        input_shape=(224, 224, 3),
        include_top=False,
        weights="imagenet"
    )
    base_model.trainable = False
    
    x = base_model(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(256, activation="relu")(x)
    
    # float32 output for numerical stability with mixed precision
    outputs = layers.Dense(NUM_CLASSES, activation="softmax", dtype='float32')(x)
    
    model = tf.keras.Model(inputs, outputs)
    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-3),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

early_stop = callbacks.EarlyStopping(monitor="val_accuracy", patience=3, restore_best_weights=True)
reduce_lr = callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.2, patience=2, min_lr=1e-6)

print("\n🚀 Super Fast Training Starting (Dono GPUs Active)...")
model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5,
    callbacks=[early_stop, reduce_lr]
)

# Phase 2: Fine-Tuning
with strategy.scope():
    base_model.trainable = True
    for layer in base_model.layers[:-30]:
        layer.trainable = False
    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-5),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

print("\n🔥 Fine-Tuning Phase Starting...")
model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=[early_stop, reduce_lr]
)

# Save Final
model.save("/kaggle/working/model_4_mobilenet_finetuned.keras")
print("\n✅ Training Successfully Completed!")
```

```python
# =====================================================================
# 🚀 STEP 3 (FIXED): Pure Float32 Export for Android (No Flex Ops Error)
# =====================================================================
import tensorflow as tf
from tensorflow.keras import layers
import os

print("🔄 Converting to Pure Float32 Mobile TFLite...")

# 1. Force global policy back to pure float32
tf.keras.mixed_precision.set_global_policy('float32')

# 2. Trained model load karein
keras_path = "/kaggle/working/model_4_mobilenet_finetuned.keras"
trained_model = tf.keras.models.load_model(keras_path)
num_classes = trained_model.output_shape[-1]
print(f"📦 Loaded trained model with {num_classes} classes.")

# 3. Clean Float32 MobileNetV2 inference architecture banayein
inputs = tf.keras.Input(shape=(224, 224, 3), name="input_image", dtype=tf.float32)
x = layers.Rescaling(1./127.5, offset=-1, dtype=tf.float32)(inputs)

# Base backbone
base_mobilenet = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights=None
)
x = base_mobilenet(x, training=False)
x = layers.GlobalAveragePooling2D(dtype=tf.float32)(x)
x = layers.BatchNormalization(dtype=tf.float32)(x)
x = layers.Dense(256, activation="relu", dtype=tf.float32)(x)
outputs = layers.Dense(num_classes, activation="softmax", dtype=tf.float32)(x)

clean_model = tf.keras.Model(inputs=inputs, outputs=outputs)

# 4. Trained model ke weights safely transfer karein
# (Sirf parameter layers ke weights transfer honge, float16 cast ops hat jayenge)
trained_weights = [w.numpy() for w in trained_model.weights if "random_" not in w.name.lower()]
clean_model.set_weights(trained_weights)
print("✅ Weights successfully transferred to clean Float32 model!")

# 5. Convert to Android TFLite
converter = tf.lite.TFLiteConverter.from_keras_model(clean_model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]  # 4x Quantization for Android NNAPI
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS]

tflite_model = converter.convert()

tflite_path = "/kaggle/working/model_4_mobilenet_finetuned.tflite"
with open(tflite_path, "wb") as f:
    f.write(tflite_model)

size_mb = os.path.getsize(tflite_path) / (1024 * 1024)
print(f"\n🎉 SUCCESS! Android TFLite binary generated: {tflite_path} ({size_mb:.2f} MB)")
print("👉 Ab Kaggle Output panel se 'model_4_mobilenet_finetuned.tflite' aur 'labels.txt' download kar sakte hain!")
```
